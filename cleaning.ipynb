```python
import pandas as pd
import os

# Input path
input_path = 'uncleaned_csvs/Wednesday-workingHours.pcap_ISCX.csv'

# 1. Setup the output directory
# This creates a folder named 'Cleaned_CSVs' one level up from your script
output_dir = './Cleaned_CSVs'
os.makedirs(output_dir, exist_ok=True)

# 2. Define the new output filename
base_name = os.path.basename(input_path)
output_name = os.path.join(output_dir, f'cleaned_{base_name}')

print(f"Loading {input_path}...")
df = pd.read_csv(input_path)

# Clean column names
df.columns = df.columns.str.strip()

# Feature selection
selected_features = [
    'Destination Port', 'Flow Duration', 'Flow Bytes/s', 'Flow Packets/s',
    'Total Fwd Packets', 'Total Backward Packets', 'Total Length of Fwd Packets',
    'Total Length of Bwd Packets', 'Fwd Packet Length Mean', 'Fwd Packet Length Std',
    'Bwd Packet Length Mean', 'Bwd Packet Length Std', 'Packet Length Mean',
    'Packet Length Std', 'Flow IAT Mean', 'Flow IAT Std', 'Fwd IAT Mean',
    'Fwd IAT Std', 'Bwd IAT Mean', 'Bwd IAT Std', 'Fwd Packets/s',
    'Bwd Packets/s', 'Fwd Header Length', 'Bwd Header Length',
    'SYN Flag Count', 'ACK Flag Count', 'PSH Flag Count', 'RST Flag Count',
    'FIN Flag Count', 'Average Packet Size', 'Avg Fwd Segment Size',
    'Avg Bwd Segment Size', 'Subflow Fwd Packets', 'Subflow Fwd Bytes',
    'Subflow Bwd Packets', 'Subflow Bwd Bytes', 'Init_Win_bytes_forward',
    'Init_Win_bytes_backward', 'Active Mean', 'Active Std', 'Idle Mean',
    'Idle Std', 'Label'
]

# Process and Clean
existing_features = [col for col in selected_features if col in df.columns]
df_selected = df[existing_features].copy() # Using copy() avoids SettingWithCopyWarning

# Strip whitespace from Label values
if 'Label' in df_selected.columns:
    df_selected['Label'] = df_selected['Label'].astype(str).str.strip()

# Replace infinity and fill NaNs
df_selected = df_selected.replace([float('inf'), -float('inf')], 0)
df_selected = df_selected.fillna(0)

# Save to the new directory
df_selected.to_csv(output_name, index=False)
print(f"Successfully cleaned. New dataset saved to: {output_name}")
```
